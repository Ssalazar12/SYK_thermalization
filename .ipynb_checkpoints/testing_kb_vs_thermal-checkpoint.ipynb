```python
def plot_imagine(g_matrix, Tmin, Tmax,bwidth=7,bheight=5,title_size=23):
    
    # makes a contour plot
    
    fig, ax = plt.subplots(1,1,figsize=(bwidth*1.1,bwidth))
    
    valnorm = abs(g_matrix.imag.min())
    im = ax.contourf(g_matrix.imag, extent=[Tmin, Tmax,Tmin,Tmax], 
                 cmap='seismic', vmin=-valnorm,vmax=valnorm)
    # plot guiding lines for the quadrants
    ax.scatter(0,0, c='grey',s=50)
    ax.axvline(0, c='grey',linewidth=2, linestyle='dashed')
    ax.axhline(0, c='grey',linewidth=2, linestyle='dashed')
    
    ax.set_title(r'$\rm{Im}[g(t_1,t_2)]$', fontsize=title_size-3)
    ax.set_ylabel(r'$t_2$', fontsize=26)
    ax.set_xlabel(r'$t_1$', fontsize=26)
    
    divider = make_axes_locatable(ax)
    cax = divider.append_axes('right', size='5%', pad=0.05)
    fig.colorbar(im, cax, shrink=0.9)

    plt.tight_layout()
    
    return fig, ax

def anti_diagonal(arr, anti_index):
    # main anti-diagonal is at anti_index=0
    # negative anti_index move the antidiagonal to the right (past the main anti)
    # positive anti_index move it to the left  (before the main anti)
    anti_arr = np.fliplr(arr)
    dindices = kb.kth_diag_indices(arr,anti_index)
    
    return anti_arr[dindices]
```

# Replacing stationary KB solution in the thermal equation

THis means that we grab the (Wigner transformed) $g(\mathcal{T}_{max},t)$ and plug it back into the thermal equation $\frac{d^2g(t)}{dt^2} = -2\mathcal{J}_q^2 e^{g(t)} - \mathcal{J}_2^2$ to see if it is also a solution. This is the reverse of next section **Replacing Thermal solution in the KB equation**, both of this should prove that there are two solutions for the KB/thermal equations.

```python
def ode(t,u1,u2,Jq,J2):
    ode_1 = u2
    ode_2 = -2*(Jq**2)*np.exp(u1) - 2*(J2**2)
    return np.array([ode_1,ode_2])

def solution_to_thermal(time_steps, g0, dg0, Dt, Jq,J2,gkb):
    # solves the thermalized equations using Runge Kutta 4. Uses scipy ODE
    # times_steps (int): number of steps the simulations will have
    # g0, dg0 (complex float): initial condition for the function and derivative respectively
    # Dt (float): step size for each timestep

    # initialize the initial conditions
    t0 = 0.0
    y0 = g0
    dy0 = dg0
    N = time_steps
    h = Dt
    tf = h*N # final time

    # For RK4 we transform to a system com 1st order coupled ODE's
    t = np.empty(N)
    y = np.empty(N,dtype=complex) # the actual function we want
    u = np.empty(N,dtype=complex) # the derivative
    
    # here the idea is g'' = -2*(J**2)*np.exp(g) - 2*J2**2
    # so we transform to: u=y' and y=g s.t. u'=  -2*(J**2)*np.exp(g) - 2*J2**2 and y'=u

    t[0] = t0
    y[0] = y0; u[0] = dy0 
    # RK4
    for i in range(0,N-1,1):
        # Estimate derivative at the start of the interval
        k11 = h*ode(t[i],gkb[i],u[i],Jq,J2)[0]
        k12 = h*ode(t[i],gkb[i],u[i],Jq,J2)[1]

        # Estimate derivatie at the middle of the interval
        k21 = h*ode(t[i]+(h/2),gkb[i]+(k11/2),u[i]+(k12/2),Jq,J2)[0]
        k22 = h*ode(t[i]+(h/2),gkb[i]+(k11/2),u[i]+(k12/2),Jq,J2)[1]
        
        # Again Estimate derivatie at the middle of the interval
        k31 = h*ode(t[i]+(h/2),gkb[i]+(k21/2),u[i]+(k22/2),Jq,J2)[0]
        k32 = h*ode(t[i]+(h/2),gkb[i]+(k21/2),u[i]+(k22/2),Jq,J2)[1]

        # Estimate derivatie at the end of the interval
        k41 = h*ode(t[i]+h,gkb[i]+k31,u[i]+k32, Jq,J2)[0]
        k42 = h*ode(t[i]+h,gkb[i]+k31,u[i]+k32, Jq,J2)[1]
        
        # calculate the solutions at the next time step
        y[i+1] = y[i] + ((k11+2*k21+2*k31+k41)/6)
        u[i+1] = u[i] + ((k12+2*k22+2*k32+k42)/6)
        t[i+1] = t[i] + h
        
    return y, u, t

print()
```

## LHS vs RHS

So first, dicretize the derivative in the LHS as $$LHS = \frac{-g_{i+2} + 16g_{i+1} - 30g_{i}+16_{i-1}-g_{i-2}}{12(\Delta t)^2}$$ 

And the RHS as $$RHS = -2\mathcal{J}_q^2 e^{g_i} - 2\mathcal{J}_2^2$$

We will now plug the wigner kb solution as $g_i$ and if it is indeed a soltion to the ODE then we get that
$$RHS = LHS$$


```python
def LHS(g, dg0, DT):
    # discretization of the derivative on the LHS
    # g: array containing the solution in wigner coordinates
    # dg0: initial condition for the derivative of g wrt t1-t2 = t
    dg_lhs = np.zeros(len(g)-1,dtype=complex)
    
    dg_lhs[0] = gprime
    # forward difference for the boundary
    dg_lhs[1] = (g[1] - g[0])/DT
    for i in range(2,len(dg_lhs)-2):
        dg_lhs[i] = (-g[i+2] + 16*g[i+1]-30*g[i]+16*g[i-1]-g[i-2])/(12*DT**(2))
        
    return dg_lhs
    
    
def RHS(g):
    # calculates the RHS of the stationary equation
    # g: array containing the solution in wigner coordinates
    return -2*(Jq[-1]**2)*np.exp(g[1:]) - 2*J2[-1]**2
    
```

```python
dg_lhs =  LHS(g_tau, gprime, Dt)
g_RHS = RHS(g_tau)

print("MSE = ", np.mean((dg_lhs - g_RHS)**2))

ind = 0
fig, ax  = plt.subplots(1,2, figsize=(bwidth*2,bheight+1))
fig.suptitle(r'$\mathcal{{J}}_q = {}$, $\mathcal{{J}}_2 = {}$, $\mathcal{{T}}=\mathcal{{T}}_{{\rm max}}$'.format(Jq[-1],J2[-1]))
ax[0].plot(t_[1+ind:],dg_lhs[ind:].real,linewidth=5, label='LHS')
ax[0].plot(t_[1:],g_RHS.real,linewidth=4,linestyle='dashed',label='RHS')
ax[0].legend()
#ax[0].set_ylim(-5,0.2)
ax[0].set_ylabel('Real')
ax[0].set_xlabel('t')

ax[1].plot(t_[1+ind:],dg_lhs[ind:].imag,linewidth=5)
ax[1].plot(t_[1:],g_RHS.imag,linewidth=4,linestyle='dashed')
#ax[1].set_ylim(-0.6,0.6)
ax[1].set_ylabel('Imaginary')
ax[1].set_xlabel('t')
plt.tight_layout()
plt.savefig(fig_route+'leftright_test_Jtwo{:.3f}.pdf'.format(J2[-1]),
            bbox_inches='tight',dpi=300)

plt.figure()
plt.plot(t_[1+ind:],dg_lhs[ind:].imag,linewidth=5)
plt.plot(t_[1:],g_RHS.imag,linewidth=4,linestyle='dashed')
plt.ylim(-0.00002,0.00002)
```

**remeber that the max $\tau$ is only $t_{max}/2$**

# Spectral Function

For the Neq case we use the wigner-ville transform. This basically means that we fourier transform the diagonals so we get a $\tau$ dependent frequency analysis.

In equilibrium according to Eberlein we have 

$A(\omega) = G^{>} (\omega) (1+ e^{-\beta\omega})$ 

Maybe if I cannot get rid off the $q$ factors I can use the distributions used by Haldar et al. in page 4 for figure 2.


```python
from scipy import fft
from scipy import signal
```

### For the real time solutions

```python
# g_full = np.concatenate((np.flip( np.conjugate(g_tau)), g_tau ))
g_full = g_thermal
normalize = len(g_full)/2
keldysh_g = fft.fft(g_full - np.conj(g_full))
keldysh_g = fft.fftshift(keldysh_g)

# compute the frequencies
freqs = fft.fftfreq(len(g_full), d=dt)
freqs = fft.fftshift(freqs)
# plot the absolute value of the Fourier Transform
plt.figure()
plt.plot(freqs, np.abs(keldysh_g)/normalize)


g_full = g_tau
keldysh_g = fft.fft(g_full - np.conj(g_full))
keldysh_g = fft.fftshift(keldysh_g)

# compute the frequencies
freqs = fft.fftfreq(len(g_full), d=dt)
freqs = fft.fftshift(freqs)
# plot the absolute value of the Fourier Transform
plt.plot(freqs, np.abs(keldysh_g )/normalize)


plt.title('Magnitude of Fourier Transform')
plt.xlabel('Frequency')
plt.ylabel('Magnitude')
plt.xlim(-0.5,0.5)
#plt.ylim(-0.1,0.5)
plt.yscale('log')

plt.show()
```

```python
# for the retarded green's function since we have a step function we fill
# negative times with 0
# gt_r = np.concatenate((np.zeros(len(g_tau)), g_tau ))
gt_r = g_thermal
# calculate  the delta
sigma = 0.01
delta = np.exp(-freqs**2 / (2 * sigma**2)) / (sigma * np.sqrt(2*np.pi))
fourier_part = fft.fft(gt_r + np.conj(gt_r)) 
fourier_part = fft.fftshift(fourier_part)

spectral_func = ( 4j*np.pi*delta + 2j*fourier_part ).imag

plt.figure()
plt.plot(freqs,np.abs(spectral_func))
plt.xlim(-0.1,0.1)
plt.yscale('log')
```
